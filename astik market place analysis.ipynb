# Supermarket Sales Analysis
### Exploratory Data Analysis (EDA) — Project 1

**Aim:** To study the supermarket transactions recorded at three branches (A, B and C) and find out how customers behave, how sales perform, which payment methods are preferred and how each product line performs.

**Python libraries:** Pandas, Matplotlib and Seaborn.

## 1. Importing the libraries and loading the data

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

                   
df = pd.read_csv('/content/supermarket.csv')

                             
display(df.head())
```

## 2. Initial exploratory data analysis (EDA)

### About the dataset
This dataset records supermarket transactions from three different branches, and every row stands for a single invoice (transaction). Its columns cover the branch, city, customer type, gender, product line, quantity, payment method, sales amount, gross income and the rating given by the customer.

```python
display(df.head())

                
display(df.tail())

                            
print("Dataset shape (rows, columns):", df.shape)

                        
print("Total number of values in the dataset:", df.size)

                             
df.info()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())

                                          
display(df.describe())
```

**Observation:** There are 1,000 transactions and 17 columns in the dataset. Before moving on, review the missing-value counts and the data types shown above. For every numerical column, the summary table lists the mean, standard deviation, minimum, maximum and quartile values.

## 3. Part 2 — Analysis questions and answers

### Q1. Do more male or female customers shop at the supermarket?

```python
gender_count = df['Gender'].value_counts()
print(gender_count)

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Gender')
plt.title('Transaction Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Transaction Count')
plt.show()
```

**Observation:** Look at the number of transactions made by male and female customers. Whichever group has the larger count made more transactions in this dataset.

### Q2. How are customer ratings distributed, and how skewed is the distribution?

```python
print("Mean customer rating:", df['Rating'].mean())
print("Rating skewness:", df['Rating'].skew())

plt.figure(figsize=(8, 4))
sns.histplot(data=df, x='Rating', kde=True)
plt.title('Spread of Customer Ratings')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.show()
```

**Interpreting skewness:** When the value is near 0, the distribution is roughly symmetrical. A positive value means the data is skewed to the right, and a negative value means it is skewed to the left.

### Q3. How do total (aggregate) sales compare across the branches?

```python
branch_sales = df.groupby('Branch')['Total'].sum().sort_values(ascending=False)
print(branch_sales)

plt.figure(figsize=(7, 4))
sns.barplot(x=branch_sales.index, y=branch_sales.values)
plt.title('Branch-wise Total Sales')
plt.xlabel('Branch')
plt.ylabel('Total Sales')
plt.show()
```

**Observation:** The branch with the largest total contributes the most to sales. The `Total` column holds the sales value of each transaction, with tax included.

### Q4. Which payment method do customers use most often?

```python
payment_count = df['Payment'].value_counts()
print(payment_count)

plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='Payment', order=payment_count.index)
plt.title('Popularity of Payment Methods')
plt.xlabel('Payment Method')
plt.ylabel('Transaction Count')
plt.show()
```

### Q4b. How are the payment methods distributed within each branch?

```python
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x='Payment', hue='Branch')
plt.title('Branch-wise Distribution of Payment Methods')
plt.xlabel('Payment Method')
plt.ylabel('Transaction Count')
plt.legend(title='Branch')
plt.show()
```

### Q5. Is customer rating influenced by gross income?

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Rating', y='gross income')
plt.title('Customer Rating vs Gross Income')
plt.xlabel('Customer Rating')
plt.ylabel('Gross Income')
plt.show()

print("Correlation between gross income and rating:", df['gross income'].corr(df['Rating']))
```

**Observation:** Read the scatter plot together with the correlation value. A correlation near zero suggests little or no linear relationship between the two. Keep in mind that correlation alone does not prove that one variable causes the other.

### Q6. Compare the distribution of ratings across the branches.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Branch', y='Rating')
plt.title('Customer Ratings Across Branches')
plt.xlabel('Branch')
plt.ylabel('Rating')
plt.show()

print(df.groupby('Branch')['Rating'].describe())
```

### Q7. Which branch earns the highest profit?

```python
branch_profit = df.groupby('Branch')['gross income'].sum().sort_values(ascending=False)
print(branch_profit)

plt.figure(figsize=(7, 4))
sns.barplot(x=branch_profit.index, y=branch_profit.values)
plt.title('Branch-wise Gross Income')
plt.xlabel('Branch')
plt.ylabel('Total Gross Income')
plt.show()
```

**Observation:** Here, profitability is measured using gross income, since it is the gross income value recorded against every transaction in the dataset.

### Q8. How is gross income related to gender?

```python
gender_income = df.groupby('Gender')['gross income'].sum()
print(gender_income)

plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x='Gender', y='gross income')
plt.title('Spread of Gross Income by Gender')
plt.xlabel('Gender')
plt.ylabel('Gross Income')
plt.show()
```

### Q9. Does gross income show any trend over time?

```python
df['Date'] = pd.to_datetime(df['Date'])

                            
daily_income = df.groupby('Date')['gross income'].sum().sort_index()

plt.figure(figsize=(12, 5))
sns.lineplot(x=daily_income.index, y=daily_income.values)
plt.title('Gross Income Over Time (Daily)')
plt.xlabel('Date')
plt.ylabel('Gross Income')
plt.xticks(rotation=45)
plt.show()

                                                  
monthly_income = df.groupby(df['Date'].dt.to_period('M'))['gross income'].sum()
print(monthly_income)
```

**Observation:** Day-to-day changes can be seen in the line chart, while the monthly totals give an easier-to-read summary of the overall trend across the three months in this dataset.

### Q10. Which product line brings in the highest income?

```python
product_income = df.groupby('Product line')['gross income'].sum().sort_values(ascending=False)
print(product_income)

plt.figure(figsize=(10, 5))
sns.barplot(x=product_income.values, y=product_income.index)
plt.title('Product Line-wise Gross Income')
plt.xlabel('Gross Income')
plt.ylabel('Product Line')
plt.show()
```

### Q11. How do male and female customers spend across the different product categories?

```python
gender_spending = df.pivot_table(
    index='Product line',
    columns='Gender',
    values='Total',
    aggfunc='sum'
)
print(gender_spending)

gender_spending.plot(kind='bar', figsize=(11, 5))
plt.title('Male vs Female Spending Across Product Lines')
plt.xlabel('Product Line')
plt.ylabel('Total Spending')
plt.xticks(rotation=35, ha='right')
plt.legend(title='Gender')
plt.tight_layout()
plt.show()
```

**Observation:** For each product line, compare the totals for males and females. The gender with the bigger value recorded more spending in that category.

### Q12. How many products do customers buy?

```python
print("Total number of products sold:", df['Quantity'].sum())
print("Mean quantity bought per transaction:", df['Quantity'].mean())

plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='Quantity')
plt.title('Transaction Count by Quantity Purchased')
plt.xlabel('Quantity per Transaction')
plt.ylabel('Transaction Count')
plt.show()
```

### Q13. On which day of the week are sales the highest?

```python
df['Weekday'] = df['Date'].dt.day_name()

weekday_sales = df.groupby('Weekday')['Total'].sum()
weekday_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
weekday_sales = weekday_sales.reindex(weekday_order)
print(weekday_sales.sort_values(ascending=False))

plt.figure(figsize=(10, 5))
sns.barplot(x=weekday_sales.index, y=weekday_sales.values)
plt.title('Weekday-wise Total Sales')
plt.xlabel('Day of the Week')
plt.ylabel('Total Sales')
plt.xticks(rotation=30)
plt.show()
```

**Observation:** The day showing the largest total sales is the best-performing sales day in this dataset. Note that this is based on the value of sales rather than just the count of transactions.

### Q14. At what hour of the day is the supermarket busiest?

```python
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M')
df['Hour'] = df['Time'].dt.hour

hour_transactions = df['Hour'].value_counts().sort_index()
print("Number of transactions in each hour:")
print(hour_transactions)
print("Hour with the most transactions:", hour_transactions.idxmax())

plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Hour', order=sorted(df['Hour'].unique()))
plt.title('Hourly Transaction Count')
plt.xlabel('Hour of the Day')
plt.ylabel('Transaction Count')
plt.show()
```

### Q15. Which product line deserves the supermarket's focus?

```python
product_summary = df.groupby('Product line').agg(
    total_sales=('Total', 'sum'),
    gross_income=('gross income', 'sum'),
    quantity_sold=('Quantity', 'sum'),
    average_rating=('Rating', 'mean')
).sort_values('gross_income', ascending=False)

display(product_summary)
```

**Observation:** Before recommending any product line, look at gross income, total sales, quantity sold and average rating together. The product line that earns the most gross income is a good place to start, but customer demand and stock levels should be taken into account as well.

### Q16. Which city should the supermarket expand into, and which products should it focus on?

```python
city_sales = df.groupby('City')['Total'].sum().sort_values(ascending=False)
city_product_income = df.pivot_table(
    index='City',
    columns='Product line',
    values='gross income',
    aggfunc='sum'
)

print("City-wise total sales:")
print(city_sales)
print("\nGross income for each city and product line:")
display(city_product_income)

plt.figure(figsize=(8, 4))
sns.barplot(x=city_sales.index, y=city_sales.values)
plt.title('City-wise Total Sales')
plt.xlabel('City')
plt.ylabel('Total Sales')
plt.show()
```

**Observation:** The city that currently has the highest sales is a sensible option to look into further; however, this dataset on its own cannot prove that it is the ideal place for a new branch. Factors such as population, rent, competition and running costs need to be considered before any actual expansion decision is made. The city-by-product table can be used to spot the product lines that do well in each city.
